#Manual

```python
from collections import defaultdict

def map_reduce(data, mapper, reducer):
    mapped = []
    for record in data:
        mapped.extend(mapper(record))

    shuffle = defaultdict(list)
    for key, value in mapped:
        shuffle[key].append(value)

    reduced = {}
    for key, values in shuffle.items():
        reduced[key] = reducer(key, values)

    return reduced


# Q1 Word Count
print("Q1:", map_reduce(
    ["hadoop is fast", "hadoop is scalable"],
    lambda l: [(w,1) for w in l.split()],
    lambda k,v: sum(v)
))


# Q2 Character Count
print("Q2:", map_reduce(
    ["big data"],
    lambda l: [(c,1) for c in l if c!=" "],
    lambda k,v: sum(v)
))


# Q3 Avg Word Length
print("Q3:", map_reduce(
    ["data science data big data"],
    lambda l: [(w,(len(w),1)) for w in l.split()],
    lambda k,v: sum(x[0] for x in v)/sum(x[1] for x in v)
))


# Q4 Global Avg
print("Q4:", map_reduce(
    ["hadoop mapreduce spark"],
    lambda l: [("global",(len(w),1)) for w in l.split()],
    lambda k,v: sum(x[0] for x in v)/sum(x[1] for x in v)
))


# Q5 Top 5 Words
text = ["this is big data big data hadoop spark data"]
res5 = map_reduce(text,
    lambda l: [(w.lower(),1) for w in l.split()],
    lambda k,v: sum(v)
)
print("Q5:", res5)
print("Top 5:", sorted(res5.items(), key=lambda x: x[1], reverse=True)[:5])


# Q6 Avg Marks
print("Q6:", map_reduce(
    ["A 80","B 70","A 90","B 60","A 100"],
    lambda l: [(l.split()[0],(int(l.split()[1]),1))],
    lambda k,v: sum(x[0] for x in v)/sum(x[1] for x in v)
))


# Q7 Avg Salary + Highest Dept
res7 = map_reduce(
    ["HR 50000","IT 70000","HR 60000","IT 80000"],
    lambda l: [(l.split()[0],(int(l.split()[1]),1))],
    lambda k,v: sum(x[0] for x in v)/sum(x[1] for x in v)
)
print("Q7:", res7)
print("Highest Dept:", max(res7, key=res7.get))


# Q8 Avg Temperature
print("Q8:", map_reduce(
    ["New York,38","London,29","Tokyo,35","New York,32","Delhi,45","Ambala,35"],
    lambda l: [(l.split(",")[0],(int(l.split(",")[1]),1))],
    lambda k,v: sum(x[0] for x in v)/sum(x[1] for x in v)
))


# Q9 Sample (instead of large dataset)
print("Q9:", map_reduce(
    ["India,30","India,32","USA,20","USA,22"],
    lambda l: [(l.split(",")[0],(float(l.split(",")[1]),1))],
    lambda k,v: sum(x[0] for x in v)/sum(x[1] for x in v)
))
```
